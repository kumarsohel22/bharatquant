```python
import os
from dotenv import load_dotenv
from huggingface_hub import login
from datasets import load_dataset

# Load environment variables
load_dotenv()

# Get Hugging Face token
HF_TOKEN = os.getenv("HF_TOKEN")

if not HF_TOKEN:
    raise ValueError("HF_TOKEN not found in .env file")

# Authenticate with Hugging Face 
login(token=HF_TOKEN, add_to_git_credential=False)

print("Hugging Face authentication successful!")

# Import dataset
dataset = load_dataset(
    "xxparthparekhxx/indian-stock-market-minute-data",
    split="day",
    streaming=True,
    token=HF_TOKEN
)

print(dataset.features)
```

```python
print("Dataset name:", dataset.info.dataset_name)

print("Number of records:",
      f"{dataset.info.splits['minute'].num_examples:,}")

print("Dataset size (GB):",
      round(dataset.info.splits['minute'].num_bytes / (1024**3), 2))

print("Download size (GB):",
      round(dataset.info.download_size / (1024**3), 2))
```
