```python
import os
from dotenv import load_dotenv
from huggingface_hub import login
from datasets import load_dataset

# Load environment variables
load_dotenv()

# Get Hugging Face token
HF_TOKEN = os.getenv("HF_TOKEN")

if not HF_TOKEN:
    raise ValueError("HF_TOKEN not found in .env file")

# Authenticate with Hugging Face 
login(token=HF_TOKEN, add_to_git_credential=False)

print("Hugging Face authentication successful!")

# Import dataset
dataset = load_dataset(
    "xxparthparekhxx/indian-stock-market-minute-data",
    "minute",
    split="train",
    streaming=True,
    token=HF_TOKEN
)

print(dataset.features)
```

```python
print("Dataset name:", dataset.info.dataset_name)

print("Dataset features:", dataset.features)

print("Available splits:", dataset.info.splits)
```

```python
if dataset.info.splits and "train" in dataset.info.splits:
    split_info = dataset.info.splits["train"]

    print("Number of records:", split_info.num_examples)

    print(
        "Dataset size (GB):",
        round(split_info.num_bytes / (1024**3), 2)
    )
else:
    print("Split metadata is not available.")
```

```python
print(dataset.info)
print(dataset.info.splits)
print(dataset.info.download_size)
print(dataset.info.dataset_size)
```

```python
count = sum(1 for _ in dataset)

print(f"Total records: {count:,}")
```

```python

# 4. Dataset information
print("\n========== DATASET INFORMATION ==========")

print("Dataset name:", dataset.info.dataset_name)
print("Dataset features:", dataset.features)

# 5. Split metadata
print("\n========== SPLIT INFORMATION ==========")

splits = dataset.info.splits

if splits and "train" in splits:
    split_info = splits["train"]

    print("Records:", split_info.num_examples)
    print("Size (GB):", round(split_info.num_bytes / (1024**3), 2))
else:
    print("Record count and size unavailable in streaming metadata.")

# 6. Download metadata
print("\n========== DOWNLOAD INFORMATION ==========")

download_size = dataset.info.download_size

if download_size is not None:
    print("Download size (GB):",
          round(download_size / (1024**3), 2))
else:
    print("Download size unavailable.")

# 7. Sample records
print("\n========== FIRST 5 RECORDS ==========")

for i, record in enumerate(dataset.take(5)):
    print(record)
```
